```python
import pandas as pd
import json

# ArXiv verisi genelde JSON formatındadır.
data = []
with open('/content/drive/MyDrive/arxiv-metadata-oai-snapshot.json', 'r') as f:
    for line in f:
        item = json.loads(line)
        # Sadece Bilgisayar Bilimi (cs) kategorisini seçelim
        if 'cs.' in item['categories']:
            data.append({
                'id': item['id'],
                'title': item['title'],
                'abstract': item['abstract']
            })
        if len(data) >= 10000: # İlk etapta 10.000 makale
            break

df = pd.DataFrame(data)
df.to_csv('arxiv_filtered.csv', index=False)
```

```python
import pandas as pd
import pickle
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sentence_transformers import SentenceTransformer
from rank_bm25 import BM25Okapi

# 1. Veriyi Yükle
df = pd.read_csv('arxiv_filtered.csv')
abstracts = df['abstract'].fillna("").tolist()

# 2. TF-IDF Hazırla ve Kaydet
tfidf_vectorizer = TfidfVectorizer(stop_words='english', max_features=10000)
tfidf_matrix = tfidf_vectorizer.fit_transform(abstracts)
with open('tfidf_model.pkl', 'wb') as f: pickle.dump(tfidf_vectorizer, f)
with open('tfidf_matrix.pkl', 'wb') as f: pickle.dump(tfidf_matrix, f)

# 3. BM25 Hazırla ve Kaydet
tokenized_corpus = [doc.lower().split(" ") for doc in abstracts]
bm25_model = BM25Okapi(tokenized_corpus)
with open('bm25_model.pkl', 'wb') as f: pickle.dump(bm25_model, f)

# 4. MPNet Vektörlerini (Embeddings) Oluştur ve Kaydet (İsim hatası düzeltildi)
model_b = SentenceTransformer('all-mpnet-base-v2')
embeddings_b = model_b.encode(abstracts, show_progress_bar=True)
np.save('embeddings_bi.npy', embeddings_b)

print("Tebrikler! Çevrimdışı hazırlık tamamlandı. Arama matrisleri diske yazıldı.")
```

```python
import chromadb
import numpy as np
import pandas as pd

# 1. Verileri ve Vektörleri Yükle
df = pd.read_csv('/content/arxiv_filtered.csv')
# ChromaDB'ye MPNet (model B) embeddings'lerini yüklemek için 'embeddings_b.npy' kullanıyoruz.
# Daha önce ChromaDB, MiniLM (model A) embeddings'leri ile doldurulmuştu. Bu değişiklik,
# ChromaDB'deki embeddings'leri MPNet ile senkronize edecektir.
embeddings = np.load('/content/embeddings_bi.npy')

# 2. ChromaDB İstemcisini Başlat (Veriler 'db_arxiv' klasörüne kaydedilecek)
client = chromadb.PersistentClient(path="/content/db_arxiv")

# 3. Bir Koleksiyon (Tablo gibi düşün) Oluştur veya Mevcudu Temizle
# Koleksiyonu temizlemek, yeni embeddings ile doldurmak için en iyi yoldur.
collection_name = "arxiv_papers"
if client.list_collections(): # Koleksiyonlar var mı kontrol et
    for col in client.list_collections():
        if col.name == collection_name:
            print(f"Mevcut '{collection_name}' koleksiyonu siliniyor...")
            client.delete_collection(name=collection_name)
            break

collection = client.get_or_create_collection(name=collection_name)
print(f"'{collection_name}' koleksiyonu oluşturuldu/temizlendi.")

# 4. Verileri Veritabanına Ekle (Batch boyutu hatasını çözmek için parçalara ayır)
# Not: ChromaDB metinleri, vektörleri ve metadata'yı aynı anda tutar
batch_size = 5000  # Maksimum batch boyutunun altında bir değer seç

for i in range(0, len(df), batch_size):
    batch_df = df.iloc[i:i + batch_size]
    batch_embeddings = embeddings[i:i + batch_size]

    collection.add(
        embeddings=batch_embeddings.tolist(),  # Numpy dizisini listeye çeviriyoruz
        documents=batch_df['abstract'].tolist(), # Orijinal metinler
        metadatas=[{"title": t} for t in batch_df['title'].tolist()], # Ek bilgiler
        ids=batch_df['id'].astype(str).tolist() # Benzersiz kimlikler
    )

print(f"Toplam {collection.count()} makale veritabanına eklendi!")
```

```python
import pandas as pd
import numpy as np
import time
import pickle
from sklearn.feature_extraction.text import TfidfVectorizer
from sentence_transformers import SentenceTransformer, CrossEncoder
from rank_bm25 import BM25Okapi
from sklearn.metrics.pairwise import cosine_similarity

# ============================================================
# 1. VERİ HAZIRLIĞI VE İNDEKSLEME (OFFLINE PHASE)
# ============================================================
def prepare_system(df):
    print("Sistem hazırlanıyor (Vektörleştirme ve İndeksleme)...")
    abstracts = df['abstract'].fillna("").tolist()

    # Model A: TF-IDF
    tfidf_vectorizer = TfidfVectorizer(stop_words='english', max_features=10000)
    tfidf_matrix = tfidf_vectorizer.fit_transform(abstracts)

    # Model B: BM25
    tokenized_corpus = [doc.lower().split(" ") for doc in abstracts]
    bm25_model = BM25Okapi(tokenized_corpus)

    # Model C: MPNet (Bi-Encoder)
    model_bi = SentenceTransformer('all-mpnet-base-v2')
    embeddings_bi = model_bi.encode(abstracts, show_progress_bar=True)

    # Model D: Cross-Encoder
    model_cross = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2')

    # Kayıt İşlemleri
    with open('tfidf_model.pkl', 'wb') as f: pickle.dump(tfidf_vectorizer, f)
    with open('tfidf_matrix.pkl', 'wb') as f: pickle.dump(tfidf_matrix, f)
    with open('bm25_model.pkl', 'wb') as f: pickle.dump(bm25_model, f)
    np.save('embeddings_bi.npy', embeddings_bi)

    return tfidf_vectorizer, tfidf_matrix, bm25_model, model_bi, embeddings_bi, model_cross

# ============================================================
# 2. PERFORMANS DEĞERLENDİRME FONKSİYONU
# ============================================================
def calculate_ndcg(found_index, k=5):
    if found_index is None or found_index >= k:
        return 0.0
    dcg = 1.0 / np.log2(found_index + 2)
    idcg = 1.0 / np.log2(1 + 1)
    return dcg / idcg

def run_comprehensive_evaluation(df, tfidf_v, tfidf_m, bm25, model_bi, embed_bi, model_cross, n_tests=200, alpha=0.3, top_k=10):
    test_samples = df.sample(n_tests, random_state=42)
    models = ["Model A (TF-IDF)", "Model B (BM25)", "Model C (MPNet)", "Model D (Cross-Only)", "Hybrid (Önerilen)"]
    stats = {m: {"R@5": 0, "P@5": 0, "NDCG@5": 0, "MRR": 0, "Time": 0} for m in models}

    print(f"[{n_tests} Sorgu] Makale formüllerine tam uyumlu performans analizi başlatıldı...\n")
    abstracts = df['abstract'].fillna("").tolist()

    for idx, row in test_samples.iterrows():
        query = row['title']
        true_id = idx
        query_tokens = query.lower().split()

        # --- 1. TF-IDF ---
        t0 = time.time()
        sim_a = (tfidf_v.transform([query]) * tfidf_m.T).toarray().flatten()
        top_a = np.argsort(sim_a)[-5:][::-1].tolist()
        stats["Model A (TF-IDF)"]["Time"] += (time.time() - t0)

        # --- 2. BM25 ---
        t1 = time.time()
        bm25_scores = bm25.get_scores(query_tokens)
        top_bm25 = np.argsort(bm25_scores)[-5:][::-1].tolist()
        stats["Model B (BM25)"]["Time"] += (time.time() - t1)

        # --- 3. MPNet (Bi-Encoder) ---
        t2 = time.time()
        q_vec = model_bi.encode([query], show_progress_bar=False)
        sim_b = cosine_similarity(q_vec, embed_bi).flatten()
        top_bi = np.argsort(sim_b)[-5:][::-1].tolist()
        stats["Model C (MPNet)"]["Time"] += (time.time() - t2)

        # --- 4. Model D (Cross-Only) ---
        t3 = time.time()
        c_pairs = [[query, abstracts[i]] for i in top_bi]
        c_scores = model_cross.predict(c_pairs)
        top_cross_only = [top_bi[i] for i in np.argsort(c_scores)[::-1]]
        stats["Model D (Cross-Only)"]["Time"] += (time.time() - t3 + (time.time() - t2))

        # --- 5. Hybrid (Makale Formüllü: Normalizasyon + Alpha + Rerank) ---
        t4 = time.time()
        # Bütün dokümanlar için ham skorları al (BM25 ve MPNet)
        raw_bm25_scores = bm25.get_scores(query_tokens)
        raw_mpnet_sims = cosine_similarity(q_vec, embed_bi).flatten()

        # Min-Max Normalizasyonu (Makale Bölüm III-C uyarınca)
        bm25_norm = (raw_bm25_scores - np.min(raw_bm25_scores)) / (np.max(raw_bm25_scores) - np.min(raw_bm25_scores) + 1e-9)
        mpnet_norm = (raw_mpnet_sims - np.min(raw_mpnet_sims)) / (np.max(raw_mpnet_sims) - np.min(raw_mpnet_sims) + 1e-9)

        # Ağırlıklı Hibrit Puanlama (Alpha = 0.3)
        hybrid_scores = (alpha * bm25_norm) + ((1 - alpha) * mpnet_norm)

        # En yüksek puanlı ilk 10 adayı seç (Stage 1)
        hybrid_candidates = np.argsort(hybrid_scores)[-top_k:][::-1].tolist()

        # Seçilen 10 adayı Cross-Encoder ile derinlemesine incele (Stage 2)
        h_pairs = [[query, abstracts[i]] for i in hybrid_candidates]
        h_scores = model_cross.predict(h_pairs)

        # Cross skorlarına göre en iyi ilk 5'i filtrele
        h_idx = np.argsort(h_scores)[::-1]
        top_hybrid = [hybrid_candidates[i] for i in h_idx][:5]

        # Süre hesaplama (Tüm adımların toplamı)
        stats["Hybrid (Önerilen)"]["Time"] += (time.time() - t4)

        # --- METRİKLERİ HESAPLA ---
        all_results = [top_a, top_bm25, top_bi, top_cross_only, top_hybrid]

        for m_name, t_list in zip(models, all_results):
            if true_id in t_list:
                rank_idx = t_list.index(true_id)
                rank = rank_idx + 1

                stats[m_name]["R@5"] += 1
                stats[m_name]["P@5"] += 1/5
                stats[m_name]["MRR"] += 1/rank
                stats[m_name]["NDCG@5"] += calculate_ndcg(rank_idx, k=5)

        current_count = list(test_samples.index).index(idx) + 1
        if current_count <= 5:
            print(f"\n>>> TEST SORGUSU {current_count}: {query[:70]}...")
            print(f"Hedef Makale ID: {true_id}")
            for m_name, t_list in zip(models, all_results):
                status = "BULDU" if true_id in t_list else " BULAMADI"
                top_titles = [df.iloc[i]['title'][:40] + "..." for i in t_list[:3]]
                print(f"  [{m_name:18}] {status} | İlk 3 Sonuç: {top_titles}")
            print("-" * 80)

    # Nihai Raporu Oluştur
    report = []
    for m in models:
        report.append({
            "Model": m,
            "Recall@5": f"{(stats[m]['R@5']/n_tests)*100:.1f}%",
            "Prec@5": f"{(stats[m]['P@5']/n_tests)*100:.1f}%",
            "NDCG@5": f"{stats[m]['NDCG@5']/n_tests:.3f}",
            "MRR": f"{stats[m]['MRR']/n_tests:.3f}",
            "Latency (ms)": f"{(stats[m]['Time']/n_tests)*1000:.1f}ms"
        })

    return pd.DataFrame(report)

# ============================================================
# 3. ANA AKIŞ (EXECUTION)
# ============================================================
df = pd.read_csv('arxiv_filtered.csv')

# 1. Adım: Çevrimdışı Hazırlık (Dosyalar diskte yoksa çalıştır, varsa doğrudan yükleyebilirsin)
tfidf_v, tfidf_m, bm25, model_bi, embed_bi, model_cross = prepare_system(df)

# 2. Adım: Makale Parametreleriyle Değerlendirme (alpha=0.3, top_k=10)
final_report = run_comprehensive_evaluation(df, tfidf_v, tfidf_m, bm25, model_bi, embed_bi, model_cross, n_tests=200, alpha=0.3, top_k=10)

# 3. Adım: Sonuçları Yazdır
print("\n" + "="*85)
print(f"{'MAKALE FORMÜLLERİNE UYUMLU FİNAL PERFORMANS RAPORU':^85}")
print("="*85)
print(final_report.to_string(index=False))
print("="*85)
```
